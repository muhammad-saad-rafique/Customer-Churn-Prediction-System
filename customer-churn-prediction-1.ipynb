```python
df_test = df_test.drop(index=199295)
```

```python
df_train["Gender"] = df_train["Gender"].map({
    "Male" : 1,
    "Female" : 0
})
df_test["Gender"] = df_test["Gender"].map({
    "Male" : 1,
    "Female" : 0
})
```

```python
df_train = pd.get_dummies(
    df_train,
    columns = ["Subscription Type", "Contract Length"],
    dtype = int
)
df_test = pd.get_dummies(
    df_test,
    columns = ["Subscription Type", "Contract Length"],
    dtype = int
)
```

```python
df_train.head()
```

```python
df = pd.concat([df_train, df_test], ignore_index=True)
```

```python
X = df.drop(columns = ["CustomerID", "Churn"])
Y = df["Churn"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split (
    X, 
    Y,
    test_size=0.2,
    stratify = Y,
    random_state = 42
)

print (X_train.shape)
print (X_test.shape)
print (Y_train.shape)
print (Y_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=100,
    n_jobs=-1,
    random_state=42
)

rf_model.fit(X_train, Y_train)

import joblib

joblib.dump(rf_model, "random_forest_model.pkl")
```

```python
rf_results = rf_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

accuracy = accuracy_score(Y_test, rf_results)
precision = precision_score(Y_test, rf_results)
recall = recall_score(Y_test, rf_results)
F1_score = f1_score(Y_test, rf_results)
matrix = confusion_matrix(Y_test, rf_results)

print("Accuracy: ", accuracy)
print("Precision: ", precision)
print("Recall: ", recall)
print("F1 Score: ", F1_score)
print("Confusion Matrix: ", matrix)
```
